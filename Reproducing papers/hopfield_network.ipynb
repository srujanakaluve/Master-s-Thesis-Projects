## With Asynchronous update and early stopping

```python
# generate random patterns
p, N = 13, 100
patterns = hopfield.get_random_patterns(p, N)

# train weight matrix
weights = hopfield.get_weights(N, patterns)

# choose and corrupt pattern
corrupted_state = hopfield.corrupt_pattern(patterns[:, 1], noise=0.1)

# run recall
iterations = 4
state, energy_list, overlap_list = hopfield.recall(corrupted_state, weights, N, iterations, recall_type="async", 
                                                   track_energy=True, track_overlap=True, pattern=patterns[:,1])

# plot energy
plt.scatter(list(range(1, ITERATIONS+1)), energy_list)
plt.ylabel("Energy")
plt.xlabel("Iterations")
plt.show()

# get overlap
print("The overlap between the original and retrieved pattern is m =", hopfield.get_overlap(state, patterns[:,1]))

plt.scatter(list(range(1, ITERATIONS+1)), overlap_list)
plt.ylabel("Overlap")
plt.xlabel("Iterations")
plt.show()
```

## With Synchronous Update and early stopping

```python
p, N = 10, 100
patterns = hopfield.get_random_patterns(p, N)

# train weight matrix
weights = hopfield.get_weights(N, patterns)

# choose and corrupt pattern
corrupted_state = hopfield.corrupt_pattern(patterns[:, 7], noise=0.1)

# run recall
iterations = 15
state, energy_list, overlap_list = hopfield.recall(corrupted_state, weights, N, iterations, 
                                                   recall_type="sync", track_energy=True, track_overlap=True, pattern=patterns[:,7])

# plot energy
plt.scatter(list(range(1, iterations+1)), energy_list)
plt.ylabel("Energy")
plt.xlabel("Iterations")
plt.show()

plt.scatter(list(range(1, iterations+1)), overlap_list)
plt.ylabel("Overlap")
plt.xlabel("Iterations")
plt.show()
```

```python
N = 708
pattern_range = np.arange(1, 800, 25)
iterations = 200
pattern_range, distances_sync = hopfield.capacity(N, iterations, pattern_range)
```

```python
plt.scatter(pattern_range, distances_sync)
plt.xlabel('Number of stored patterns')
plt.ylabel('MI')
plt.show()
```
